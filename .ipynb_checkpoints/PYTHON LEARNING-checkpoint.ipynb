```python
fruits=["a", "b", "c", "d", "e", "f"]
for a in fruits:
    if a=="b":
        print("found")
    elif a=="c":
        print("good")
    else:
        print("bad")
```

# FUNCTIONS

```python
lambda_fun=lambda a,b:(a+b, a*b)
type(lambda_fun)
```

```python
lambda_fun(10, 20)
```

```python
def named_state(a, b):
    if a==2 and b==3:
        print("hello")
        print("hero")
        print("lal")
    else:
        print("false case !!!!")

named_state(2, 3)
```

```python
named_state(3, 4)
```

```python
def addition(a, b):
    return a+b

addition(3, 4)
```

```python
def substraction(a, b):
    print(abs(a-b));

substraction(34, 5)
```

```python
def my_sum(*args):
    return sum(args)

my_sum(1, 2,3,4, 5, 6, 7, 8)
```

```python
def key_value_pair(**kvargs):
    print(kvargs)
    for k,v in kvargs.items():
        print(k)
        print(v)

key_value_pair(name="ravi", age=35, loc="delhi")
```

```python
def outside():
    print("outside")
    def inside():
        print("inside")
    inside()
outside()
```
